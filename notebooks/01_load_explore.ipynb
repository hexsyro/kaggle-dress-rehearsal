## 📓 Notebook 01: Load & Explore

**Skills practiced:** Loading and quickly cleaning a tabular dataset, preparing `X` and `y` without splitting yet.

**Goal:** Prepare one clean, encoded feature set and save it. The train/test split happens later, on purpose.

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
```

```python
df=pd.read_csv('../data/raw/Telco_Customer_Churn.csv').dropna(subset=['TotalCharges'])
```

```python
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
df["TotalCharges"] = df["TotalCharges"].fillna(0)
print(df["TotalCharges"].isna().sum())
```

```python
df.Churn=df.Churn.map({'Yes':1, 'No':0})
```

```python
df.info()
```

```python
x=df.drop(columns=['customerID','Churn'])
y=df['Churn']
```

```python
cats=['gender','Partner','Dependents','PhoneService','MultipleLines','InternetService','OnlineSecurity','OnlineBackup','DeviceProtection','TechSupport','StreamingTV','StreamingMovies','Contract','PaperlessBilling','PaymentMethod']

ohe = OneHotEncoder(sparse_output=False, handle_unknown="ignore", drop=None)
ohe.fit(x[cats])
   
x_enc=pd.DataFrame(ohe.transform(x[cats]), columns=ohe.get_feature_names_out())
x_enc = x.join(x_enc).drop(columns=cats)
```

```python
x.to_csv('../data/processed/x_unenc.csv',index=False)
y.to_csv('../data/processed/y.csv',index=False)
x_enc.to_csv('../data/processed/x.csv',index=False)
```
